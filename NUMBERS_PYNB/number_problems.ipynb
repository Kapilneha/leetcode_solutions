PRINT PRIME 1 T0 N

```python
def is_prime(n):
    if n < 2:
        return False

    i = 2

    while i * i <= n:
        if n % i == 0:
            return False
        i += 1

    return True


def print_primes(n):
    for num in range(2, n + 1):
        if is_prime(num):
            print(num, end=" ")


n = int(input("Enter n: "))
print_primes(n)
```

ARMSTRONG

```python
def is_armstrong(num):
    original = num
    digits = len(str(num))
    total = 0

    while num > 0:
        digit = num % 10
        total += digit ** digits
        num //= 10

    return total == original


num = int(input("Enter a number: "))

print(is_armstrong(num))
```

FACTORIAL USING RECURSION

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


n = int(input("Enter a number: "))

print(factorial(n))
```

FIBONACCI

```python
def fibonacci(n):
    a = 0
    b = 1

    for i in range(n):
        print(a, end=" ")

        a, b = b, a + b


n = int(input("Enter number of terms: "))
fibonacci(n)
```
